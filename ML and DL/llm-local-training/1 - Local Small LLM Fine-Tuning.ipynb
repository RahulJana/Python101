# Local Small LLM Fine-Tuning

This notebook is a lightweight entry point for the local training pipeline in this folder.

Recommended order:

1. Resolve the project directory and confirm the paths.
2. Check that CUDA is available.
3. Install the dependencies from `requirements.txt` after installing a CUDA-enabled PyTorch build.
4. Start with `configs/smollm2_135m_rtx3060_6gb.json`.
5. Train, evaluate, and chat with the adapter using the generated commands.


````python
from pathlib import Path
import os
import sys


def find_project_dir(start: Path) -> Path:
    markers = (
        Path("scripts/train.py"),
        Path("scripts/evaluate.py"),
        Path("scripts/chat.py"),
        Path("configs/smollm2_135m_rtx3060_6gb.json"),
    )

    for candidate in [start, *start.parents]:
        if all((candidate / marker).exists() for marker in markers):
            return candidate

    nested_candidate = start / "ML and DL" / "0-Frameworks" / "PyTorch" / "llm-local-training"
    if all((nested_candidate / marker).exists() for marker in markers):
        return nested_candidate

    for candidate in start.rglob("llm-local-training"):
        if candidate.is_dir() and all((candidate / marker).exists() for marker in markers):
            return candidate

    raise FileNotFoundError("Could not find the llm-local-training project directory from the current notebook session.")


PROJECT_DIR = find_project_dir(Path.cwd()).resolve()
os.chdir(PROJECT_DIR)

TRAIN_SCRIPT = PROJECT_DIR / "scripts" / "train.py"
EVAL_SCRIPT = PROJECT_DIR / "scripts" / "evaluate.py"
CHAT_SCRIPT = PROJECT_DIR / "scripts" / "chat.py"
CONFIG_PATH = PROJECT_DIR / "configs" / "smollm2_135m_rtx3060_6gb.json"
CHECKPOINT_DIR = PROJECT_DIR / "outputs" / "smollm2_135m_python101" / "final_checkpoint"
REQUIREMENTS_FILE = PROJECT_DIR / "requirements.txt"

print(f"Python executable: {sys.executable}")
print(f"Project directory: {PROJECT_DIR}")
print(f"Working directory: {Path.cwd()}")
print(f"Config exists: {CONFIG_PATH.exists()}")
print(f"Train script exists: {TRAIN_SCRIPT.exists()}")
print(f"Requirements file exists: {REQUIREMENTS_FILE.exists()}")
````

````python
import torch

print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    total_vram_gb = torch.cuda.get_device_properties(0).total_memory / 1024**3
    print(f"GPU memory: {total_vram_gb:.2f} GB")
````

## Install dependencies

Run the following commands in the same environment as this notebook kernel. Install a CUDA-enabled PyTorch build first, then install the local requirements:

```powershell
pip install torch torchvision torchaudio
pip install -r "C:\\full\\path\\to\\requirements.txt"
```


````python
def quote(path: Path) -> str:
    return f'"{path}"'


python_executable = quote(Path(sys.executable))
requirements_command = f"{python_executable} -m pip install -r {quote(REQUIREMENTS_FILE)}"
train_command = f"{python_executable} {quote(TRAIN_SCRIPT)} --config {quote(CONFIG_PATH)}"
eval_command = (
    f"{python_executable} {quote(EVAL_SCRIPT)} "
    f"--config {quote(CONFIG_PATH)} "
    f"--checkpoint-dir {quote(CHECKPOINT_DIR)}"
)
chat_command = (
    f"{python_executable} {quote(CHAT_SCRIPT)} "
    f"--config {quote(CONFIG_PATH)} "
    f"--checkpoint-dir {quote(CHECKPOINT_DIR)} "
    '--instruction "Explain Python decorators in simple words."'
)

print(requirements_command)
print(train_command)
print(eval_command)
print(chat_command)
````
