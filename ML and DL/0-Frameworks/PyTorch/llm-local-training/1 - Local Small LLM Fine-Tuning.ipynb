# Local Small LLM Fine-Tuning

This notebook is a lightweight entry point for the local training pipeline in this folder.

Recommended order:

1. Check that CUDA is available.
2. Install the dependencies from `requirements.txt` after installing a CUDA-enabled PyTorch build.
3. Start with `configs/smollm2_135m_rtx3060_6gb.json`.
4. Train, evaluate, and chat with the adapter using the scripts in `scripts/`.


````python
import torch

print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    total_vram_gb = torch.cuda.get_device_properties(0).total_memory / 1024**3
    print(f"GPU memory: {total_vram_gb:.2f} GB")
````

## Install dependencies

Run the following commands in an activated environment. Install a CUDA-enabled PyTorch build first, then install the local requirements:

```powershell
pip install torch torchvision torchaudio
pip install -r requirements.txt
```


````python
from pathlib import Path

config_path = Path("./configs/smollm2_135m_rtx3060_6gb.json")
train_command = f'python "./scripts/train.py" --config "{config_path}"'
eval_command = (
    'python "./scripts/evaluate.py" '
    '--config "./configs/smollm2_135m_rtx3060_6gb.json" '
    '--checkpoint-dir "./outputs/smollm2_135m_python101/final_checkpoint"'
)
chat_command = (
    'python "./scripts/chat.py" '
    '--config "./configs/smollm2_135m_rtx3060_6gb.json" '
    '--checkpoint-dir "./outputs/smollm2_135m_python101/final_checkpoint" '
    '--instruction "Explain Python decorators in simple words."'
)

print(train_command)
print(eval_command)
print(chat_command)
````
